# Participant Dictionary

```python
import pandas as pd
import polars as pl
from pathlib import Path

BRONZE = Path("../datalake/bronze/dictionaries/participant_dictionary")

# Listar o que temos
for f in sorted(BRONZE.iterdir()):
    size = f.stat().st_size / 1024
    print(f"  {f.name:50s} {size:>10.1f} KB")
```

```python
df_csv = pd.read_csv(BRONZE / "participant_d040722.csv")

print(f"Shape: {df_csv.shape}")
print(f"\nColunas ({len(df_csv.columns)}):")
print(df_csv.columns.tolist())
print(f"\nTipos:\n{df_csv.dtypes}")
print(f"\nPrimeiras linhas:")
df_csv.head()
```

```python
df_csv.columns.tolist()
```

```python
# O .sas é um script SAS com PROC FORMAT — vamos ler como texto pra entender a estrutura
sas_path = BRONZE / "participant.sas_formats.d040722.sas"

with open(sas_path, "r", encoding="latin-1") as f:
    sas_content = f.read()

# Mostrar as primeiras 200 linhas
for i, line in enumerate(sas_content.splitlines()[:200]):
    print(f"{i+1:4d} | {line}")
```

```python
# O CSV tem os dados com códigos numéricos
# O SAS formats tem os mapeamentos código → label
# Exemplo: no CSV, gender=1. No SAS formats, 1="Male"

# Vamos ver um exemplo concreto
print("── CSV: valores únicos de 'gender' ──")
print(df_csv["gender"].value_counts())

print("\n── CSV: valores únicos de 'race' ──")
print(df_csv["race"].value_counts())

print("\n── CSV: valores únicos de 'cigsmok' ──")
print(df_csv["cigsmok"].value_counts())
```

```python
print(f"Total participantes: {len(df_csv)}")
print(f"PIDs únicos: {df_csv['pid'].nunique()}")
print(f"\n── Nulls por coluna ──")
null_pct = (df_csv.isnull().sum() / len(df_csv) * 100).sort_values(ascending=False)
print(null_pct[null_pct > 0].to_string())
```

```python
# Carregar os patient_ids que já temos no Gold
import duckdb

con = duckdb.connect("../datalake/nlst.duckdb", read_only=True)

# Pacientes no masterfile (que conectam patologia ↔ lung cancer)
mf_pids = set(con.sql("SELECT DISTINCT patient_id FROM masterfile").fetchdf()["patient_id"].astype(str))

# Pacientes no lung_cancer
lc_pids = set(con.sql("SELECT DISTINCT patient_id FROM lung_cancer").fetchdf()["patient_id"].astype(str))

con.close()

# Pacientes no participant CSV
part_pids = set(df_csv["pid"].astype(str))

print(f"Participant CSV:     {len(part_pids)} pacientes")
print(f"Masterfile (Gold):   {len(mf_pids)} pacientes")
print(f"Lung Cancer (Gold):  {len(lc_pids)} pacientes")

print(f"\nParticipant ∩ Masterfile:    {len(part_pids & mf_pids)}")
print(f"Participant ∩ Lung Cancer:   {len(part_pids & lc_pids)}")
print(f"Participant ∩ Ambos:         {len(part_pids & mf_pids & lc_pids)}")
```

```python
# Pegar um pid que existe em tudo
common_pids = list(part_pids & mf_pids & lc_pids)
if common_pids:
    sample_pid = common_pids[0]
    print(f"Paciente exemplo: {sample_pid}")
    print(f"\n── Dados no Participant CSV ──")
    display(df_csv[df_csv["pid"] == int(sample_pid)])
```
